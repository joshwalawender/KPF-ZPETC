```python
from pathlib import Path
import numpy as np
from astropy import stats
from astropy.time import Time
from astropy.table import Table
from kpf_etc.etc import kpf_photon_noise_estimate

import matplotlib.pyplot as plt

data_file = 'data/Feb2026_anonymized.csv'
all_data = Table.read(data_file, format='ascii.csv')
len(all_data)
```

```python
# KPFERA 1.0 is before SM1
SM1start = Time('2024-02-03')
SM1end = Time('2024-02-23')
# KPFERA 2.0 is between SM1 and SM2
SM2start = Time('2024-10-31')
SM2end = Time('2024-11-20')
# KPFERA 2.5/2.6 is between SM2 and SM3
SM3start = Time('2025-03-28')
SM3end = Time('2025-04-23')
# KPFERA 3.0 is between SM3 and SM4
SM4start = Time('2025-08-29')
SM4end = Time('2025-10-27')
# KPFERA 4.0 is after SM4

dataByEra = {'1.0': all_data[(all_data['MJD-OBS'] > 60200) & (all_data['MJD-OBS'] < SM1start.mjd)],
             '2.0': all_data[(all_data['MJD-OBS'] > SM1end.mjd) & (all_data['MJD-OBS'] < SM2start.mjd)],
             '2.5': all_data[(all_data['MJD-OBS'] > SM2end.mjd) & (all_data['MJD-OBS'] < SM3start.mjd)],
             '3.0': all_data[(all_data['MJD-OBS'] > SM3end.mjd) & (all_data['MJD-OBS'] < SM4start.mjd)],
             '4.0': all_data[(all_data['MJD-OBS'] > SM4end.mjd)],
            }
for era in dataByEra.keys():
    print(era, len(dataByEra[era]))
```
